```python
def clean_time_str(val):
    if pd.isna(val):
        return val
    try:
        v = float(val)
        if 0 <= v <= 1:
            total_sec = int(v * 86400)
            return f"{total_sec // 3600:02d}:{(total_sec % 3600) // 60:02d}"
    except ValueError:
        pass
    return str(val).strip()

accident_all['เวลา'] = accident_all['เวลา'].apply(clean_time_str)
accident_all['ชั่วโมง'] = accident_all['เวลา'].str.extract(r'^(\d{1,2}):')[0]
accident_all['ชั่วโมง'] = pd.to_numeric(accident_all['ชั่วโมง'], errors='coerce')
accident_all['ชั่วโมง'] = accident_all['ชั่วโมง'].fillna(-1).astype(int)
```

```python
def get_time_period(ชั่วโมง):
    if pd.isna(ชั่วโมง):
        return 'ไม่ระบุ'
    elif 0 <= ชั่วโมง < 4:
        return 'ดึก (00:00 - 03:59)'
    elif 4 <= ชั่วโมง < 8:
        return 'เช้ามืด (04:00 - 07:59)'
    elif 8 <= ชั่วโมง < 12:
        return 'เช้า (08:00 - 11:59)'
    elif 12 <= ชั่วโมง < 16:
        return 'บ่าย (12:00 - 15:59)'
    elif 16 <= ชั่วโมง < 20:
        return 'เย็น (16:00 - 19:59)'
    else:
        return 'กลางคืน (20:00 - 23:59)'
```
