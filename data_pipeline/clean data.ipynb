```python
import pandas as pd
```

```python
df2021 = pd.read_csv("accident2021.csv", encoding="utf-8-sig")
df2022 = pd.read_csv("accident2022.csv", encoding="utf-8-sig")
df2023 = pd.read_csv("accident2023.csv", encoding="utf-8-sig")
df2024 = pd.read_csv("accident2024.csv", encoding="utf-8-sig")
df2025 = pd.read_csv("accident2025.csv", encoding="utf-8-sig")
```

```python
accident_all['วันที่เกิดเหตุ'] = pd.to_datetime(
    accident_all['วันที่เกิดเหตุ'], 
    format='mixed', 
    errors='coerce'
)
```

```python
accident_all['เดือน'] = accident_all['วันที่เกิดเหตุ'].dt.month
```

```python
thai_days = {
    "Monday": "จันทร์", "Tuesday": "อังคาร", "Wednesday": "พุธ",
    "Thursday": "พฤหัสบดี", "Friday": "ศุกร์", "Saturday": "เสาร์", "Sunday": "อาทิตย์"
}
accident_all['วันในสัปดาห์'] = accident_all['วันที่เกิดเหตุ'].dt.day_name().map(thai_days)
```

```python
try:
    accident_all['ชั่วโมง'] = pd.to_datetime(accident_all['เวลา'].astype(str), errors='coerce').dt.hour
except:
    accident_all['ชั่วโมง'] = accident_all['เวลา'].astype(str).str.extract(r'(\d+)')[0]
    accident_all['ชั่วโมง'] = pd.to_numeric(accident_all['ชั่วโมง'], errors='coerce')
```

```python
print("เช็กจำนวนค่าว่างหลังแก้ไข:")
print(accident_all[['วันที่เกิดเหตุ', 'เดือน', 'วันในสัปดาห์', 'ชั่วโมง']].isnull().sum())
```

```python
accident_all['LATITUDE'] = pd.to_numeric(accident_all['LATITUDE'], errors='coerce')
accident_all['LONGITUDE'] = pd.to_numeric(accident_all['LONGITUDE'], errors='coerce')
```

```python
valid_lat = (accident_all['LATITUDE'] >= 5.0) & (accident_all['LATITUDE'] <= 21.0)
valid_long = (accident_all['LONGITUDE'] >= 97.0) & (accident_all['LONGITUDE'] <= 106.0)
```

```python
#การกรองช่วง Lat 5.0 - 21.0 และ Long 97.0 - 106.0 เป็นวิธีมาตรฐานในงาน Data Science เพื่อไม่ให้หมุดเบี้ยวเวลาปักบน Interactive Map
```

```python
accident_all = accident_all[valid_lat & valid_long].copy()
```

```python
numeric_cols = [
    'รถจักรยานยนต์', 'รถสามล้อเครื่อง', 'รถยนต์นั่งส่วนบุคคล', 'รถตู้',
    'รถปิคอัพโดยสาร', 'รถโดยสารมากกว่า4ล้อ', 'รถปิคอัพบรรทุก4ล้อ', 
    'รถบรรทุก6ล้อ', 'รถบรรทุกไม่เกิน10ล้อ', 'รถบรรทุกมากกว่า10ล้อ', 
    'รถอีแต๋น', 'รถอื่นๆ', 'คนเดินเท้า', 
    'ผู้เสียชีวิต', 'ผู้บาดเจ็บสาหัส', 'ผู้บาดเจ็บเล็กน้อย', 'รวมจำนวนผู้บาดเจ็บ'
]

#แทนที่ค่าว่าง (NaN) ด้วย 0 และแปลงเป็น integer
for col in numeric_cols:
    if col in accident_all.columns:
        accident_all[col] = accident_all[col].fillna(0).astype(int)

print("จำนวน Missing Values ในคอลัมน์ตัวเลข:")
print(accident_all[numeric_cols].isnull().sum())
```

```python
text_cols = [
    'บริเวณที่เกิดเหตุ', 'มูลเหตุสันนิษฐาน', 'ลักษณะการเกิดเหตุ', 
    'สภาพอากาศ', 'จังหวัด', 'รถคันที่1', 'สายทาง'
]

for col in text_cols:
    if col in accident_all.columns:
        accident_all[col] = accident_all[col].fillna('ไม่ระบุ').astype(str).str.strip()
```

```python
accident_all['ชั่วโมง'] = accident_all['เวลา'].astype(str).str.extract(r'(\d+)')[0]
accident_all['ชั่วโมง'] = pd.to_numeric(accident_all['ชั่วโมง'], errors='coerce')

accident_all['ชั่วโมง'] = accident_all['ชั่วโมง'].fillna(-1).astype(int)
#-1 คือการสร้าง Out-of-bound Category หรือ Sentinel Value เพื่อใช้จัดการกับ "ข้อมูลที่ไม่ทราบเวลา / ข้อมูลขยะ"
```
