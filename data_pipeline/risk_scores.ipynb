```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv('new_accident.csv')
```

```python
column_mapping = {
    'หน่วยงาน': 'agency',
    'สายทาง': 'road_name',
    'จังหวัด': 'province',
    'รถคันที่1': 'vehicle',
    'บริเวณที่เกิดเหตุ': 'location_type',
    'มูลเหตุสันนิษฐาน': 'presumed_cause',
    'ลักษณะการเกิดเหตุ': 'incident_type',
    'สภาพอากาศ': 'weather',
    'ผู้เสียชีวิต': 'deaths',
    'ผู้บาดเจ็บสาหัส': 'serious_injuries',
    'ผู้บาดเจ็บเล็กน้อย': 'minor_injuries',
    'รวมจำนวนผู้บาดเจ็บ': 'total_injuries',
    'ชั่วโมง': 'hour',
    'LATITUDE' : 'latitude',
    'LONGITUDE' : 'longitude',
    'is_dangerous_7days' : 'is_dangerouse_7days'
}
df2 = df1.rename(columns=column_mapping)
```

```python
df2.info()
```

```python
conditions = [
    df2["is_songkran"] == 1,
    df2["is_new_year"] == 1,
    df2["is_dangerouse_7days"] == 1,
    df2["is_weekend"] == 1,
]
choices = ["songkran", "new_year", "danger_7days", "weekend"]
```

```python
df2["day_type"] = np.select(conditions, choices, default="normal_day")
```

```python
df2['day_type'].unique()
```

```python
df2.to_csv('new_accident.csv', index=False, encoding='utf-8-sig')
print("บันทึกแล้ว")
```

```python
df3 = df2.copy()
```

```python
df3['risk_score'] = (
    15 + 
    (df2['deaths'] * 25) + 
    (df2['serious_injuries'] * 15) + 
    (df2['minor_injuries'] * 5)
).clip(upper=100)

def assign_risk_level(score):
    if score >= 70:
        return 'High'
    elif score >= 40:
        return 'Medium'
    else:
        return 'Low'

df3['risk_level'] = df3['risk_score'].apply(assign_risk_level)
```

```python
df3.info()
```

```python
df3.to_csv('new_accident2.csv', index=False, encoding='utf-8-sig')
print("บันทึกแล้ว")
```
