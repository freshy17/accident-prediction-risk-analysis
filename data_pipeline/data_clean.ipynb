```python
df_clean['ชั่วโมง'] = df_clean['เวลา'].astype(str).str.extract(r'(\d+)')[0]
df_clean['ชั่วโมง'] = pd.to_numeric(df_clean['ชั่วโมง'], errors='coerce')
df_clean['ชั่วโมง'] = df_clean['ชั่วโมง'].fillna(-1).astype(int)

def clean_time_str(val):
    if pd.isna(val):
        return val
    try:
        v = float(val)
        if 0 <= v <= 1:
            total_sec = int(v * 86400)
            return f"{total_sec // 3600:02d}:{(total_sec % 3600) // 60:02d}"
    except ValueError:
        pass
    return str(val).strip()

df_clean['เวลา'] = df_clean['เวลา'].apply(clean_time_str)
df_clean['ชั่วโมง'] = df_clean['เวลา'].str.extract(r'^(\d{1,2}):')[0]
df_clean['ชั่วโมง'] = pd.to_numeric(df_clean['ชั่วโมง'], errors='coerce')
df_clean['ชั่วโมง'] = df_clean['ชั่วโมง'].fillna(-1).astype(int)
```

```python
def parse_mixed_dates(val):
    try:
        num = float(val)
        if num > 1000:
            return pd.to_datetime(num, unit='D', origin='1899-12-30')
    except (ValueError, TypeError):
        pass
    
    return pd.to_datetime(val, dayfirst=True, errors='coerce')

df_clean['วันที่เกิดเหตุ'] = df_clean['วันที่เกิดเหตุ'].apply(parse_mixed_dates)
```

```python
df_clean['year'] = df_clean['วันที่เกิดเหตุ'].dt.year
df_clean['month'] = df_clean['วันที่เกิดเหตุ'].dt.month
df_clean['day'] = df_clean['วันที่เกิดเหตุ'].dt.day
df_clean['dayofweek'] = df_clean['วันที่เกิดเหตุ'].dt.dayofweek  # 0=จันทร์, 6=อาทิตย์
df_clean['is_weekend'] = df_clean['dayofweek'].isin([5, 6]).astype(int)  # 1=เสาร์-อาทิตย์, 0=จันทร์-ศุกร์

def get_time_period(ชั่วโมง):
    if pd.isna(ชั่วโมง):
        return 'ไม่ระบุ'
    elif 0 <= ชั่วโมง < 4:
        return 'ดึก (00:00 - 03:59)'
    elif 4 <= ชั่วโมง < 8:
        return 'เช้ามืด (04:00 - 07:59)'
    elif 8 <= ชั่วโมง < 12:
        return 'เช้า (08:00 - 11:59)'
    elif 12 <= ชั่วโมง < 16:
        return 'บ่าย (12:00 - 15:59)'
    elif 16 <= ชั่วโมง < 20:
        return 'เย็น (16:00 - 19:59)'
    else:
        return 'กลางคืน (20:00 - 23:59)'

df_clean['time_period'] = df_clean['ชั่วโมง'].apply(get_time_period)
df_clean[['วันที่เกิดเหตุ', 'เวลา', 'year', 'month', 'dayofweek', 'is_weekend', 'ชั่วโมง', 'time_period']].tail()
```

```python
def calculate_severity(row):
    if row['ผู้เสียชีวิต'] >= 1:
        return 'รุนแรงมาก (เสียชีวิต)'
    elif row['ผู้บาดเจ็บสาหัส'] >= 1:
        return 'รุนแรงปานกลาง (บาดเจ็บสาหัส)'
    else:
        return 'รุนแรงน้อย (บาดเจ็บเล็กน้อย/ทรัพย์สินเสียหาย)'

df_clean['severity_level'] = df_clean.apply(calculate_severity, axis=1)
```

```python
def check_new_year(date):
    if pd.isna(date): return 0
    return 1 if (date.month == 12 and date.day >= 29) or (date.month == 1 and date.day <= 4) else 0

def check_songkran(date):
    if pd.isna(date): return 0
    return 1 if (date.month == 4 and 11 <= date.day <= 17) else 0

df_clean['is_new_year'] = df_clean['วันที่เกิดเหตุ'].apply(check_new_year)
df_clean['is_songkran'] = df_clean['วันที่เกิดเหตุ'].apply(check_songkran)
df_clean['is_dangerous_7days'] = ((df_clean['is_new_year'] == 1) | (df_clean['is_songkran'] == 1)).astype(int)

print("ปีใหม่     :", df_clean['is_new_year'].sum(), "เคส")
print("สงกรานต์  :", df_clean['is_songkran'].sum(), "เคส")
print("รวม 7วันอันตราย:", df_clean['is_dangerous_7days'].sum(), "เคส")
```

```python
df_clean['day_type'] = np.select(
    [
        df_clean['is_new_year'] == 1,
        df_clean['is_songkran'] == 1,
        df_clean['is_weekend'] == 1,
    ],
    ['new_year', 'songkran', 'weekend'],
    default='normal_day'
)
print(df_clean['day_type'].value_counts())
```
